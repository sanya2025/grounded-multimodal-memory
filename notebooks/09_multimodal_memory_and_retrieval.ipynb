## Build a store and retrieve

Embed events, then run hybrid retrieval with a component breakdown.

```python
import json
from pathlib import Path
from datetime import datetime
from grounded_memory.memory.embeddings import HashingTextEmbedder
from grounded_memory.memory.events import MemoryEvent
from grounded_memory.memory.store import MemoryStore
from grounded_memory.retrieval.hybrid import HybridRetriever
seq = json.loads((Path('..')/'data'/'sequences'/'toy_sequence_01.json').read_text())
emb = HashingTextEmbedder(); store = MemoryStore()
for e in seq['events']:
    store.add(MemoryEvent(e['event_id'], datetime.fromisoformat(e['timestamp']),
        e.get('image_path',''), e['description'], e.get('entities',[]),
        embedding=emb.embed_text(e['description'])))
res = HybridRetriever(store).retrieve(query_embedding=emb.embed_text('where are the keys?'),
    query_entities={'keys'}, reference=datetime(2026,8,14,12,0), top_k=3)
for r in res: print(r.event.event_id, round(r.score,3), r.components)
```
